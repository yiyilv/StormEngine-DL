# StormEngine V9.2 — Event-aware correction
This workflow implements the original physical-event training requirement without changing frozen V9/V9.1 results. Run preflight, smoke, then pilot. Do not start the formal run until the 2018 pilot comparison is reviewed.

```python
from pathlib import Path
import subprocess, sys
ROOT = Path.cwd().resolve()
if ROOT.name == 'notebooks': ROOT = ROOT.parent
RUN_PREFLIGHT = True
RUN_SMOKE = False
RUN_PILOT = False
PROFILE = 'base'  # change to 'strong' only for the preserved second pilot
RUN_EVALUATE_PILOT = False
RUN_FORMAL = False
RUN_FINAL16_REFIT = False  # already completed; guarded against overwrite
print(ROOT)
```

```python
def run(mode):
    command = [sys.executable, '-u', str(ROOT/'scripts'/'run_v9_2_event_aware.py'), mode, '--device', 'cuda', '--profile', PROFILE]
    print(' '.join(command), flush=True)
    subprocess.run(command, cwd=ROOT, check=True)
if RUN_PREFLIGHT: run('preflight')
```

```python
if RUN_SMOKE: run('smoke')
```

```python
if RUN_PILOT: run('pilot')
```

```python
if RUN_EVALUATE_PILOT: run('evaluate-pilot')
```

```python
if RUN_FORMAL: run('train')
```
